Working with data

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

**Extract year**

```python
date['date_year'] = date['date'].dt.year
```

```python
date.head()
```

```python
date.sample(5)
```

**Extract Month**

```python
date['date_month_no'] = date['date'].dt.month
```

```python
date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()
```

```python
date.head()
```

**Extract Days**

```python
date['date_day'] = date['date'].dt.day
```

```python
date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek
```

```python
date.head()
```

```python
# day of week - name
date['date_dow_name'] = date['date'].dt.day_name()
```

```python
date.drop(columns=['product_id','city_id','orders']).head()
```

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0) # Is weekend
# date['date_is_weekend'] = ...: This part is assigning values to a new column named date_is_weekend in your date DataFrame.
# np.where(...): This is a NumPy function that performs a conditional check. It's similar to an 'if-else' statement. It takes three arguments: a condition, a value to return if the condition is true, and a value to return if the condition is false.
# date['date_dow_name'].isin(['Sunday', 'Saturday']): This is the condition. It checks each entry in the date_dow_name column to see if it is either 'Sunday' or 'Saturday'.
# 1: If the day name is 'Sunday' or 'Saturday' (i.e., the condition is true), the date_is_weekend column for that row will be set to 1.
# 0: If the day name is neither 'Sunday' nor 'Saturday' (i.e., the condition is false), the date_is_weekend column for that row will be set to 0.
```

```python
date.drop(columns=['product_id','city_id','orders']).head()
```

**Extract week of the year**

```python
date['date_week'] = date['date'].dt.isocalendar().week
```

```python
date.drop(columns=['product_id','city_id','orders']).head()
```

**Extract Quarter**

```python
date['quarter'] = date['date'].dt.quarter
```

```python
date.drop(columns=['product_id','city_id','orders']).head()
```

**Extract Semester**

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)
```

```python
date.drop(columns=['product_id','city_id','orders']).head()
```
